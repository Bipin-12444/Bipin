```python
# This is a simple Python program that provides a menu for the user
#to choose between calculating the average of numbers or concatenating strings.
#The program will continue to run until the user chooses to exit.
def calculate_average(*numbers):
    #usig the * operator to accept a variable number of arguments
    total = sum(numbers)
    average = total / len(numbers)
    return average


def concatenate_strings(**strings):
#Using the ** operator to accept a variable number of keyword arguments
    result = ""

    for value in strings.values():
        result = result + value

    return result


while True:

    print("\n===== MENU =====")
    print("1. Calculate Average")
    print("2. Concatenate Strings")
    print("3. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":

        numbers = []

        amount = int(input("How many numbers do you want to enter? "))

        for i in range(amount):
            number = float(input("Enter number: "))
            numbers.append(number)

        answer = calculate_average(*numbers)

        print("Average:", answer)

    elif choice == "2":

        strings = {}

        amount = int(input("How many strings do you want to enter? "))

        for i in range(amount):
            text = input("Enter string: ")
            strings["string" + str(i)] = text

        answer = concatenate_strings(**strings)

        print("Result:", answer)

    elif choice == "3":

        print("Exiting the program. Goodbye!")
        break

    else:

        print("Invalid choice.")

```
