```python
def add_two_numbers(number1, number2):
  return number1 + number2
def subtract_two_numbers(number1, number2):
  return number1 - number2
def multiply_two_numbers(number1, number2):
  return number1 * number2
def divide_two_numbers(number1, number2):
  return number1 / number2

def test_add_two_numbers():
  assert add_two_numbers(3, 5) == 8, 'Expected add_two_numbers(3, 5) to return 8, but received ' + str(add_two_numbers(3, 5))
  assert add_two_numbers(4, 9) == 13, 'Expected add_two_numbers(4, 9) to return 13, but received ' + str(add_two_numbers(4, 9))
  assert add_two_numbers(2, 3) == 5, 'Expected addition result to be 5, but received ' + str(add_two_numbers(2, 3))
def test_subtract_two_numbers():
  assert subtract_two_numbers(8, 4) == 4, 'Expected subtract_two_numbers(8, 4) to return 4, but received ' + str(subtract_two_numbers(8, 4))
  assert subtract_two_numbers(9, 2) == 7, 'Expected subtract_two_numbers(9, 2) to return 7, but received ' + str(subtract_two_numbers(9, 2))
  assert subtract_two_numbers(5, 2) == 3, 'Expected subtraction result to be 3, but received ' + str(subtract_two_numbers(5, 2))
def test_multiply_two_numbers():
  assert multiply_two_numbers(3, 5) == 15, 'Expected multiply_two_numbers(3, 5) to return 15, but received ' + str(multiply_two_numbers(3, 5))
  assert multiply_two_numbers(4, 9) == 36, 'Expected multiply_two_numbers(4, 9) to return 36, but received ' + str(multiply_two_numbers(4, 9))
  assert multiply_two_numbers(4, 6) == 24, 'Expected multiplication result to be 24, but received ' + str(multiply_two_numbers(4, 6))
def test_divide_two_numbers():
  assert divide_two_numbers(8, 4) == 2, 'Expected divide_two_numbers(8, 4) to return 2, but received ' + str(divide_two_numbers(8, 4))
  assert divide_two_numbers(9, 2) == 4.5, 'Expected divide_two_numbers(9, 2) to return 4.5, but received ' + str(divide_two_numbers(9, 2))
  assert divide_two_numbers(10, 2) == 5, 'Expected division result to be 5, but received ' + str(divide_two_numbers(10, 2))

test_add_two_numbers()
test_subtract_two_numbers()
test_multiply_two_numbers()
test_divide_two_numbers()
print('All arithmetic tests passed successfully!')
```
